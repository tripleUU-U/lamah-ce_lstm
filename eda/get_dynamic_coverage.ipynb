```python
import pandas as pd

from tqdm import tqdm
from pathlib import Path
```

```python
lamah_ce_path = Path("/home/wuhlmann/BA/data/raw_data/2_LamaH-CE_daily")
B_climatic_timeseries_path = lamah_ce_path / "B_basins_intermediate_all/2_timeseries/daily"
files = list(B_climatic_timeseries_path.glob("*.csv"))
```

```python
starts, ends = [], []

for file in tqdm(files):
    dynamics = pd.read_table(file, header=0, sep=";")
    dynamics = dynamics.rename(columns={"YYYY": "year", "MM": "month", "DD": "day"})

    start = pd.Timestamp(year=int(dynamics.loc[0, "year"]),
                         month=int(dynamics.loc[0, "month"]),
                         day=int(dynamics.loc[0, "day"]))

    end = pd.Timestamp(year=int(dynamics.loc[len(dynamics)-1, "year"]),
                       month=int(dynamics.loc[len(dynamics)-1, "month"]),
                       day=int(dynamics.loc[len(dynamics)-1, "day"]))

    starts.append(start)
    ends.append(end)
```
