```python
import pickle 
import pandas as pd 
import seaborn as sns

from tqdm import tqdm
from sklearn.decomposition import PCA	
```

```python
with open("/home/wuhlmann/BA/data/processed_data/cell_states/q_pred_landcover_0903_135428_cell_states.p", "rb") as f: 
	cs_dict = pickle.load(f)
```

```python
pca = PCA(2, random_state=1277)

for basin_id in tqdm(cs_dict.keys()):
	
	pca.fit(cs_dict[basin_id]["c_last"])
```

```python
pca_array = pca.transform(cs_dict["435"]["c_last"])


sns.scatterplot(x=pca_array[:,0], y=pca_array[:,1], s=0.5)
```
