```python
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.model_selection import train_test_split
```

```python
data_path = Path("/home/wuhlmann/BA/data")

coverage = pd.read_csv(
    data_path / "processed_data/eda/discharge_coverage.csv",
    header=0,
    sep=";",
    converters={"gap_dates": pd.eval}
)

coverage.set_index("basin_id", inplace=True)

valid_basin_ids = list(coverage[coverage["num_missing_days"] == 0].index)
```

```python
train_basin_ids, val_basin_ids = train_test_split(valid_basin_ids, train_size=0.8)

with open(data_path / "processed_data/test_splits/train_basin_ids.txt", "w") as f: 
    for id in sorted(train_basin_ids):
        f.write(f"{id}\n")

with open(data_path / "processed_data/test_splits/val_basin_ids.txt", "w") as f: 
    for id in sorted(val_basin_ids):
        f.write(f"{id}\n")
```
